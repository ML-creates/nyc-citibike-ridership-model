```python
X = df[
    [
        "temp_f",
        "precip_in",
        "wind_speed_knots",
        "year",
        "day_Monday",
        "day_Saturday",
        "day_Sunday",
        "day_Thursday",
        "day_Tuesday",
        "day_Wednesday"
    ]
]

y = df["num_rides"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training X:", X_train.shape)
print("Test X:", X_test.shape)
print("Training y:", y_train.shape)
print("Test y:", y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

model.fit(X_train, y_train)

print("Model fitted successfully.")
```

```python
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import numpy as np

# Make predictions
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Calculate R²
train_r2 = r2_score(y_train, y_train_pred)
test_r2 = r2_score(y_test, y_test_pred)

# Calculate MAE
mae = mean_absolute_error(y_test, y_test_pred)

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y_test, y_test_pred))

print("Training R²:", train_r2)
print("Test R²:", test_r2)
print("Test MAE:", mae)
print("Test RMSE:", rmse)
```

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coefficients
```

## Precipitation has a negative coefficient of -3,930.99, meaning that an additional inch of precipitation is associated with approximately 3,931 fewer rides,  the other variables are constant. The year coefficient is positive at 5,501.17, suggesting that ridership increased by about 5,501 rides per year after accounting for weather and day-of-week effects. Sunday has a coefficient of -7,068.65, meaning Sundays are predicted to have about 7,069 fewer rides than the reference day, Friday, when the other variables are held constant. These signs generally match the patterns observed during EDA: precipitation and wind reduce ridership, while warmer temperatures and system growth are associated with increased ridership. The temperature coefficient is worth examining further because the EDA suggested that ridership may level off or decline on extremely hot days.



```python
residuals = y_test - y_test_pred

print("Residuals calculated:", len(residuals))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(y_test_pred, residuals)

plt.axhline(y=0, linestyle="--")

plt.xlabel("Predicted Rides")
plt.ylabel("Residuals (Actual - Predicted)")
plt.title("Residuals vs. Predicted Rides")

plt.show()
```

```python
test_results = df.loc[X_test.index, ["ride_date"]].copy()

test_results["residual"] = residuals

test_results.head()
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    pd.to_datetime(test_results["ride_date"]),
    test_results["residual"]
)

plt.axhline(y=0, linestyle="--")

plt.xlabel("Ride Date")
plt.ylabel("Residuals (Actual - Predicted)")
plt.title("Residuals vs. Ride Date")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

##  The residuals versus predicted rides were mostly scattered around zero without a strong pattern, which suggests that the model is not showing a major systematic problem across its predicted values. However, the residuals plotted against ride date showed an upward trend over time. This suggests that the model tends to underpredict ridership more often in the later years. Although the model includes year as a trend feature, the growth in the Citi Bike system may not be completely captured by a simple linear trend. This is a weakness that could explain some of the model's larger errors.

```python
X_improved = df[
    [
        "temp_f",
        "temp_f_squared",
        "precip_in",
        "wind_speed_knots",
        "year",
        "day_Monday",
        "day_Saturday",
        "day_Sunday",
        "day_Thursday",
        "day_Tuesday",
        "day_Wednesday"
    ]
]

print("Improved X shape:", X_improved.shape)
```

```python
X_improved_train = X_improved.loc[X_train.index]
X_improved_test = X_improved.loc[X_test.index]

print("Improved training:", X_improved_train.shape)
print("Improved test:", X_improved_test.shape)
```

```python
improved_model = LinearRegression()

improved_model.fit(X_improved_train, y_train)

improved_y_test_pred = improved_model.predict(X_improved_test)

improved_test_r2 = r2_score(y_test, improved_y_test_pred)

print("Original Test R²:", test_r2)
print("Improved Test R²:", improved_test_r2)
```

## Model Improvement: I added temp_f_squared as one improvement to the core model because the EDA suggested that the relationship between temperature and ridership was not completely linear, particularly on very hot days. The original model had a test R² of 0.7153. After adding the squared temperature feature, the test R² increased to 0.7281. This improvement of about 0.0128 suggests that the curved temperature relationship helps the model explain additional variation in daily ridership. The improved model performed better on the test data, so I would keep this feature in the final model
