### E3 Observations

Temperature variables show the strongest relationships with daily ridership. In general, ridership increases as temperature increases. Both maximum and minimum temperature show clear positive relationships with ridership, although the observations are widely spread. The mean temperature relationship also becomes less consistent at the highest temperatures, suggesting that the relationship may not be completely linear.

Wind speed shows a weaker relationship with ridership. There appears to be some tendency for ridership to be lower at higher wind speeds, but the points are widely scattered.

Precipitation shows a less clear relationship because most observations are concentrated near zero precipitation. The value of 99.99 is clearly separated from the other observations and confirms the coded missing-value problem identified in E2.

Overall, temperature appears to have the strongest relationship with ridership, while wind speed and precipitation appear weaker.

### E5 Observations

Ridership is generally higher on weekdays than on weekends. Wednesday has the highest average daily ridership at approximately 36,265 rides, followed by Thursday and Tuesday. Sunday has the lowest average at approximately 27,261 rides, while Saturday is also lower than the weekdays.

This weekday-weekend pattern makes sense because Citi Bike is likely used for commuting and other weekday activities. The lower ridership on Saturday and Sunday may reflect fewer work and school trips. The clear difference between weekdays and weekends suggests that day of week will be an important predictor of daily ridership.

### E6 Observations

There are 186 missing calendar dates between the minimum and maximum dates in the dataset. The missing dates are not randomly distributed. Four dates are missing from January 2016 (January 23–26), followed by a continuous gap from October 1, 2016 through March 31, 2017.

The ride data shows that Citi Bike records exist immediately before and after the large gap, with rides recorded on September 30, 2016 and again beginning April 1, 2017. This suggests that Citi Bike service did not simply stop during this period.

The most likely explanation is that the missing dates came from the weather side of the INNER JOIN used to create the final dataset. Because an INNER JOIN only keeps dates that appear in both the daily rides and daily weather data, dates without a matching weather observation were excluded from the final dataset.

This is an important data quality issue because the missing dates represent a substantial portion of the date range and could affect the analysis and model if they are not recognized.
