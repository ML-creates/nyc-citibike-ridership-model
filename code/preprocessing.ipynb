```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/citibike_weather_daily.csv")
```

```python
df.dtypes
```

```python
df["ride_date"] = pd.to_datetime(df["ride_date"])
```

```python
df.dtypes
```

```python
print("Temperature 9999.9:", (df["temp_f"] == 9999.9).sum())
print("Max temperature 9999.9:", (df["max_temp_f"] == 9999.9).sum())
print("Min temperature 9999.9:", (df["min_temp_f"] == 9999.9).sum())
print("Wind 999.9:", (df["wind_speed_knots"] == 999.9).sum())
print("Precipitation 99.99:", (df["precip_in"] == 99.99).sum())
```

```python
df[df["precip_in"] == 99.99]
```

```python
df["precip_in"] = df["precip_in"].replace(99.99, np.nan)
```

```python
df[df["ride_date"] == "2016-02-11"]["precip_in"]
```

```python
precip_median = df["precip_in"].median()

precip_median
```

```python
df["precip_in"] = df["precip_in"].fillna(precip_median)
```

```python
df["precip_in"].isna().sum()
```

```python
df[df["ride_date"] == "2016-02-11"][["ride_date", "precip_in"]]
```

```python
day_dummies = pd.get_dummies(
    df["day_of_week"],
    prefix="day",
    drop_first=True,
    dtype=int
)

day_dummies.head()
```

```python
df = pd.concat([df, day_dummies], axis=1)
```

```python
df = df.drop(columns=["day_of_week"])
```

```python
df.shape
```

```python
df["year"] = df["ride_date"].dt.year
```

```python
df[["ride_date", "year"]].head()
```

```python
df["temp_f_squared"] = df["temp_f"] ** 2
```

```python
df[["temp_f", "temp_f_squared"]].head()
```

### C5 — Squared Temperature Feature

I created a squared temperature feature because E3 showed that the relationship between temperature and ridership was not completely linear, especially on the hottest days. The squared temperature term allows the linear regression model to capture some curvature in the relationship between temperature and daily ridership.

```python
df["temp_f_squared"] = df["temp_f"] ** 2
```

```python
df[["temp_f", "temp_f_squared"]].head()
```
